```python
import random

N = 4

def cost(q):
    h = 0
    for i in range(N):
        for j in range(i + 1, N):
            if abs(q[i] - q[j]) == abs(i - j):
                h += 1
    return h

def display(q):
    for i in range(N):
        for j in range(N):
            print("Q" if q[j] == i else "-", end=" ")
        print()

def hill_climbing():
    q = [3, 1, 2, 0]
    print("Initial state:", q)
    print("Initial cost:", cost(q))

    while cost(q) != 0:
        current_cost = cost(q)
        best = q[:]
        best_cost = current_cost

        for i in range(N):
            for j in range(i + 1, N):
                neighbor = q[:]
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
                neighbor_cost = cost(neighbor)

                print("Neighbor:", neighbor,
                      "Cost:", neighbor_cost)

                if neighbor_cost < best_cost:
                    best = neighbor
                    best_cost = neighbor_cost

        if best_cost >= current_cost:
            print("No better neighbor found.")
            return

        q = best
        print("\nNew state:", q)
        print("New cost:", cost(q))

    print("\nGoal state reached!")
    print("Final state:", q)
    print("Final cost:", cost(q))
    display(q)

hill_climbing()
```
