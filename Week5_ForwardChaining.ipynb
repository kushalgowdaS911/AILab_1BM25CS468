```python
facts = [
    ("Parent", "John", "Mary"),
    ("Parent", "Mary", "Sam")
]

rules = [
    ("Grandparent", "Parent", "Parent")
]

query = ("Grandparent", "John", "Sam")

def forward_chaining(facts, query):
    new_facts = True

    while new_facts:
        new_facts = False

        for rule in rules:
            for f1 in facts:
                for f2 in facts:

                    if (f1[0] == rule[1] and
                        f2[0] == rule[2] and
                        f1[2] == f2[1]):

                        new_fact = (
                            rule[0],
                            f1[1],
                            f2[2]
                        )

                        if new_fact not in facts:
                            facts.append(new_fact)
                            new_facts = True

                            print("New Fact:", new_fact)

                        if new_fact == query:
                            return True

    return query in facts


print("Initial Facts:")
for fact in facts:
    print(fact)

if forward_chaining(facts, query):
    print("\nQuery Proved: TRUE")
else:
    print("\nQuery Proved: FALSE")
```
