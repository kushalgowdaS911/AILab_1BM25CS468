```python
# 8 Puzzle Problem using Depth Limited DFS

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def print_board(state):
    for i in range(0, 9, 3):
        for value in state[i:i + 3]:
            if value == 0:
                print(" ", end=" ")
            else:
                print(value, end=" ")
        print()
    print()


def get_neighbors(state):
    zero = state.index(0)

    row = zero // 3
    col = zero % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    neighbors = []

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_zero = new_row * 3 + new_col

            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            neighbors.append(tuple(new_state))

    return neighbors


def dfs(state, goal, depth, max_depth, path, visited):

    # Goal test
    if state == goal:
        return path

    # Stop if maximum depth is reached
    if depth >= max_depth:
        return None

    visited.add(state)

    # Explore each possible move
    for next_state in get_neighbors(state):

        if next_state not in visited:

            result = dfs(
                next_state,
                goal,
                depth + 1,
                max_depth,
                path + [next_state],
                visited
            )

            if result is not None:
                return result

    return None


# ---------------- MAIN PROGRAM ----------------

print("8 PUZZLE USING DEPTH FIRST SEARCH")
print("----------------------------------")

print("\nEnter the initial state:")
print("Use 0 for blank space.")

numbers = list(map(int, input().split()))

if len(numbers) != 9:
    print("Please enter exactly 9 numbers.")
else:

    initial_state = tuple(numbers)

    print("\nInitial State:")
    print_board(initial_state)

    print("Goal State:")
    print_board(GOAL)

    # Maximum depth
    MAX_DEPTH = 30

    visited = set()

    solution = dfs(
        initial_state,
        GOAL,
        0,
        MAX_DEPTH,
        [initial_state],
        visited
    )

    if solution is not None:

        print("Goal State Found!")

        print("\nSolution Path:")

        for step, state in enumerate(solution):

            print("Step", step)
            print_board(state)

        print("Total moves:", len(solution) - 1)

    else:

        print("Goal not found within depth limit", MAX_DEPTH)
```
